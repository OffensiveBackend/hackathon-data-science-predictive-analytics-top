```python
# import pandas
import pandas as pd

# Create a DataFrame
# df = data_frame
df = pd.read_csv('./data/supermarket_data.csv')

 
# A data frame is just how the data in the dat set is presented, e.g. table, charts, pie/bar graph. 
# our dataset is a table.
# print the dataFrame.
df
```

```python
# describe the data set. Identifying patterns in our dataset and drawing conclusions.
df.describe() # Produces a multi.variate dimensioned data set
```

```python
# draw a bar graphs leveraging seaborn library
# seaborn is a library visual graphs
import seaborn as sn # seaborn is a modern improved variant of matploitlib

# import matploitlib
import matplotlib.pyplot as plt

# plot a graph (bar graph) for the gender that buys items from the supermarket
sn.countplot(x="gender", data=df, palette="dark")
plt.show()
```

```python
# plot a graph for the most liked product by customers
# source of data, naming of graph's axis
sn.countplot(x="product", data=df, palette="dark") # (x='name_of_column', data='source_of_data', pallette='give_the_colouring')
plt.xticks(rotation=-45) # optimised for human readability
# plt.show() # display the actual graph.
```

```python
"""
Create a count: 
#  1. Which gender has the highest number of costumers overall?
#  2. In which city is each gender mostcommon?
#  3. Which city has the lowest number of customers for a given gender?
#  4. What differences do u observe btwn cities?
#  5.  What does this tell u about customer distribution across cities?
#  6. What recommendation would u give to the business?
"""

# plot the graph for the customer distributionb across cities
# No. of customers based on 
sn.countplot(x="gender", hue="city", palette="Reds", data=df) # hue="on_the_basis_of?" => basis_of_grouping, e.g. group according to city
plt.show()

# we use this graph to answer these questions 
```
