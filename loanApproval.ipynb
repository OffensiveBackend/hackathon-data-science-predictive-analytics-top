```python
# library
import pandas as pd

# PROCESS OF DATA CLEANING
# target data set
df = pd.read_csv("./data/bank.csv")

# shape of data set
df
```

```python
# Inspect the data for inconsistencies
print(f"Column:{df.columns}") # Just the column_head
print(df.head()) # column_head with first 5 rows
```

```python
# Properties of our dataframe
# Information about the data set
print(f"Shape: {df.shape[0]}") # a 614 rowed table
print(f"\n Missing values per column: {df.isnull().sum()}") # Number of empty rows with respect to a given column ,e.g., column 'self_Employed' has 30 empty rows
```

```python
# PROCESS OF DATA CLEANING
# handle the missing values
before_drop = df.shape[0]

# drop the rows with the missing values in the dataset
df = df.dropna()

# also remove the duplicates from the dataset
df = df.drop_duplicates()

# shape after cleaning
after_cleaning = df.shape[0] # start at index 0, which is row
change = (before_drop - after_cleaning) # no. duplicate rows

# print the results
print(f"Shape after cleaning\n : {after_cleaning}")
print(f"Number of duplicates removed \n: {change}")
print(f"\n Data after cleaning\n {df.isnull().sum()}") # see whether data set still has empty columns

# NOTE: After cleaning, the original data set isn't altered but how the computer see it.
```
